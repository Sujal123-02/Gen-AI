```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv("weather.csv")
```

```python
df
```

```python
df["outlook"]=df["outlook"].replace({'rainy':0,'sunny':1,'overcast':2})
df["temperature"]=df["temperature"].replace({'cool':0,'mild':1,'hot':2})
df["humidity"]=df["humidity"].replace({'normal':0,'high':1})
df["windy"]=df["windy"].replace({'False':0,'True':1})
# df["play"]=df["play"].replace({'no':0,'yes':1})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score
```

```python
x=df.drop("play",axis=1)
y=df["play"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
tree_model=DecisionTreeClassifier()
tree_model.fit(X_train,y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree

```

```python
plot_tree(tree_model)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
tree=plot_tree(tree_model,feature_names=x.columns)
```

```python
y_pred = tree_model.predict(X_test)
y_pred
```

```python
print("Accuracy : ",accuracy_score(y_test,y_pred))
```
