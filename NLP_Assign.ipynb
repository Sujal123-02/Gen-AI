```python
documents = [
    "I enjoy exploring deep learning.",
"Artificial intelligence inspires innovation.",
"Machine learning transforms possibilities into reality.",
"AI is shaping the future.",
"Learning algorithms fascinate me.",
"Intelligent systems are revolutionary.",
"Exploring neural networks excites me.",
"AI makes technology smarter.",
"Machine learning unlocks hidden patterns.",
"Artificial intelligence drives progress."
]

for doc in documents:
    print(doc)
```

```python
import re

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()

    stop_words = {"i", "is", "the", "a", "an", "very"}

    words = [word for word in words if word not in stop_words]

    return words


processed_documents = []

for doc in documents:
    words = preprocess(doc)
    processed_documents.append(words)

    print("Original :", doc)
    print("Processed:", words)
    print()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

clean_text = [" ".join(words) for words in processed_documents]

vectorizer = CountVectorizer()

bow = vectorizer.fit_transform(clean_text)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBoW Matrix:")
print(bow.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(clean_text)

print("Vocabulary:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
from gensim.models import Word2Vec

word2vec_model = Word2Vec(
    sentences=processed_documents,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)

print("Vector of 'machine':")
print(word2vec_model.wv["machine"])
```

```python
print("Similar words to 'machine':")

print(word2vec_model.wv.most_similar("machine", topn=3))
```

```python
from gensim.models import FastText

fasttext_model = FastText(
    sentences=processed_documents,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)

print("Vector of 'machine':")
print(fasttext_model.wv["machine"])
```

```python
print("Vector of 'machines':")
print(fasttext_model.wv["machines"])
```

Stemming Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).




```python
from nltk.corpus import stopwords

# Example token list
tokens = ['Welcome','to','this','Generative','AI','lecture','where','we','explore','how','machines','can','learn']

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Filter out stopwords
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
words=['Welcome', 'Generative', 'AI', 'lecture', 'explore', 'machines', 'learn']

for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
### Lancaster  Stemming algorithm
from nltk.stem import LancasterStemmer
lancaster=LancasterStemmer()
for word in words:
    print(word+"---->"+lancaster.stem(word))
```

RegexpStemmer class NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example




```python
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer('ing|s$|e$|able$', min=4) 
for word in words:
    print(word+"---->"+reg_stemmer.stem(word))
```

Snowball Stemmer




```python
from nltk.stem import SnowballStemmer
snowballstemmer=SnowballStemmer('english',ignore_stopwords=False)
for word in words:
    print(word+"---->"+snowballstemmer.stem(word))
```

Wordnet Lemmatizer Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4') # Download Open Multilingual WordNet
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```
