```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
```

```python
(x_train, y_train), (x_test,y_test) = datasets.cifar10.load_data()
x_train.shape
```

```python
y_train = y_train.reshape(-1,)   #2D array can be reshaped into 1D array using reshape(-1)
y_train[:5]
```

```python
y_test = y_test.reshape(-1,)
y_test[:5]
```

```python
classes = ["airplane","automobile","bird","cat","deer","dog","frog","horse","ship","truck"]
```

```python

def plot_sample(x, y, index):
    plt.figure(figsize = (20,2))
    plt.imshow(x[index])
    plt.xlabel(classes[y[index]])
```

```python
plot_sample(x_train, y_train, 0)
```

```python
plot_sample(x_train, y_train, 1)
```

```python
plot_sample(x_train, y_train, 4)
```

```python
plot_sample(x_train, y_train, 400)
```

```python
plot_sample(x_train, y_train, 4112)
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
     
```

```python
x_train[0]
```

```python
annmodel = models.Sequential([
        layers.Flatten(input_shape=(32,32,3)),
        layers.Dense(1000, activation='relu'),
        layers.Dense(1000, activation='relu'),
        layers.Dense(10, activation='softmax')    
    ])
```

```python
annmodel.compile(optimizer='SGD',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
annmodel.fit(x_train, y_train, epochs=5)
```
